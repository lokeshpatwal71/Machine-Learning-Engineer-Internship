## 4. Introduction and Theory
**Series** = one labelled column; **DataFrame** = a table of Series sharing an index. pandas builds on NumPy, so calculations are vectorised.

* `df[col]` selects a column; `df.loc[rows, cols]` selects by label; `df.iloc[i, j]` selects by position.
* A **boolean filter** such as `df[df.x > 5]` keeps rows where the condition is True. Combine conditions with `&` and `|` (with brackets).
* **groupby** splits the data into groups, applies a function (mean, count ...) and combines the result: *split - apply - combine*.
* **Missing values** appear as `NaN`. `isna()` finds them, `dropna()` removes them and `fillna()` replaces them. The right choice depends on why the value is missing.

### Step 1: Creating a DataFrame

```python
import numpy as np
import pandas as pd

students = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meena", "Kabir"],
    "course": ["DS", "DS", "ML", "ML"],
    "marks": [88, 54, 71, 93],
})
students["passed"] = students["marks"] >= 60
display(students)
print(students.dtypes)
```

### Step 2: Load Iris and inspect it

```python
from pathlib import Path
from sklearn.datasets import load_iris

Path("../data/raw").mkdir(parents=True, exist_ok=True)
iris = load_iris(as_frame=True).frame
iris["species"] = load_iris().target_names[iris.pop("target")]
iris.columns = [c.replace(" (cm)", "").replace(" ", "_") for c in iris.columns]
print("Shape:", iris.shape)
display(iris.head())
iris.info()
```

```python
display(iris.describe().round(2))
print("Missing values:", int(iris.isna().sum().sum()), "| Duplicate rows:", int(iris.duplicated().sum()))
print(iris["species"].value_counts())
```

Iris contains 1 fully duplicated row(s). Identical measurements can legitimately occur in real data, and nothing says they are errors, so we keep them.

### Step 3: Select and filter

```python
print(iris["petal_length"].head(3).tolist())                       # one column -> Series
display(iris[["species", "petal_length"]].head(3))                  # several columns -> DataFrame
display(iris.loc[10:12, ["sepal_length", "species"]])               # by label
display(iris.iloc[0:3, 0:2])                                        # by position

big = iris[iris["petal_length"] > 6]
print("Petal length > 6 cm:", len(big), "flowers; species:", big["species"].unique().tolist())
combo = iris[(iris["species"] == "setosa") & (iris["sepal_width"] > 4)]
print("Setosa with sepal width > 4:", len(combo))
display(iris.sort_values("petal_length", ascending=False).head(3))
```

### Step 4: Group and summarise

```python
by_species = iris.groupby("species").agg(
    flowers=("species", "size"),
    mean_petal_length=("petal_length", "mean"),
    mean_petal_width=("petal_width", "mean"),
    std_sepal_length=("sepal_length", "std"),
).round(3)
display(by_species)
print("Highest mean petal length:", by_species["mean_petal_length"].idxmax())
```

### Step 5: Load Stocks.csv and find missing values

```python
import matplotlib

stocks_path = Path(matplotlib.get_data_path()) / "sample_data" / "Stocks.csv"
stocks = pd.read_csv(stocks_path, comment="#", parse_dates=["Date"]).rename(columns={"^GSPC": "SP500", "^IXIC": "NASDAQ"})
stocks.to_csv("../data/raw/stocks.csv", index=False)
print("Shape:", stocks.shape, "| from", stocks["Date"].min().date(), "to", stocks["Date"].max().date())

missing = stocks.isna().sum().sort_values(ascending=False)
display(pd.DataFrame({"missing": missing, "percent": (missing / len(stocks) * 100).round(1)}))
all_blank = stocks.drop(columns="Date").isna().all(axis=1)
print("Rows where every price is blank:", int(all_blank.sum()))
```

Some rows are completely empty (only a date). These are not real trading months, so we drop them. The remaining gaps are mostly from before a company was listed (for example Google before 2004): the price did not exist, so **we must not fill them with an invented number**.

```python
clean = stocks.loc[~all_blank].copy()
print("Rows before:", len(stocks), "| after dropping empty rows:", len(clean))
print("First valid GOOGL date:", clean.loc[clean["GOOGL"].notna(), "Date"].min().date())
print("First valid AMZN  date:", clean.loc[clean["AMZN"].notna(), "Date"].min().date())
print("Remaining missing cells:", int(clean.isna().sum().sum()))
```

### Step 6: New columns and a time-based summary

```python
tech = clean[["Date", "IBM", "AAPL", "MSFT"]].dropna().copy()
tech["year"] = tech["Date"].dt.year
tech["AAPL_return_pct"] = tech["AAPL"].pct_change() * 100        # change from the previous row
tech["MSFT_to_IBM_ratio"] = (tech["MSFT"] / tech["IBM"]).round(3)

yearly = tech.groupby("year")[["IBM", "AAPL", "MSFT"]].mean().round(2)
display(yearly.tail(5))
print("Best AAPL monthly change:", round(tech["AAPL_return_pct"].max(), 2), "% on", tech.loc[tech["AAPL_return_pct"].idxmax(), "Date"].date())
```

### Step 7: Export the results

```python
Path("../data/processed").mkdir(parents=True, exist_ok=True)
by_species.to_csv("../data/processed/iris_species_summary.csv")
yearly.to_csv("../data/processed/stocks_yearly_mean_prices.csv")
print("Saved:", sorted(p.name for p in Path("../data/processed").glob("*.csv")))
```

## 11. Observations and Analysis
* `describe()`, `value_counts()` and `isna()` give a quick health check before any analysis.
* Missing data is not always a mistake. Empty rows were dropped, but the pre-listing gaps in newer companies are left empty because filling them would invent prices.
* `groupby` with named aggregation is a clean way to build a summary table in one step.
* Filtering with `&` needs brackets around each condition; forgetting them gives an error.

**Common mistakes**
| Mistake | Fix |
|---|---|
| `df[df.a > 1 and df.b < 2]` | Use `&` with brackets: `df[(df.a > 1) & (df.b < 2)]` |
| Modifying a filtered slice (SettingWithCopy warning) | Use `.copy()` first |
| Filling missing values without thinking | Decide first why they are missing |
