## 4. Introduction and Theory
| Type | Example | Notes |
|---|---|---|
| `int` | `42` | whole numbers, any size |
| `float` | `3.14` | decimal numbers (small rounding errors are normal) |
| `str` | `"data"` | text; cannot be changed in place (immutable) |
| `bool` | `True` | a truth value; behaves like 1 and 0 in arithmetic |
| `list` | `[1, 2, 3]` | ordered, **changeable**, allows duplicates |
| `tuple` | `(1, 2, 3)` | ordered, **not changeable** |
| `set` | `{1, 2, 3}` | **unordered, no duplicates**, fast membership tests |
| `dict` | `{"a": 1}` | key-value pairs, fast lookup by key |

**Indexing** starts at 0; negative numbers count from the end (`-1` is the last item). **Slicing** `seq[start:stop:step]` includes `start` but excludes `stop`.

### Step 1: The four basic types

```python
count = 42            # int
price = 19.99         # float
subject = "Python"    # str
is_open = True        # bool

for value in (count, price, subject, is_open):
    print(f"{value!r:>10}  ->  {type(value).__name__}")

print("\nint + float gives:", type(count + price).__name__)
print("True + True =", True + True)           # bool behaves like 0 and 1
print("0.1 + 0.2 =", 0.1 + 0.2)               # floats are approximate
print("round(0.1 + 0.2, 2) =", round(0.1 + 0.2, 2))
```

### Step 2: Strings

```python
text = "  Data Science with Python  "
print(repr(text.strip()))
print(text.strip().upper())
print(text.strip().lower())
print(text.strip().replace("Python", "pandas"))
print(text.strip().split())                       # split into words
print("-".join(["data", "science"]))              # join words with a separator
print(len(text.strip()), "characters")
print("Science" in text)                           # membership test
name, marks = "Asha", 88.456
print(f"{name} scored {marks:.1f} marks")          # f-string with formatting
```

### Step 3: Lists (ordered and changeable)

```python
scores = [72, 85, 90, 66]
scores.append(78)            # add at the end
scores.insert(1, 99)         # add at position 1
scores.remove(66)            # remove the first value 66
print("List after changes:", scores)
print("Sorted:", sorted(scores), "| original untouched:", scores)
scores.sort(reverse=True)    # sorts the list itself
print("Sorted high to low in place:", scores)
print("Length:", len(scores), "| Max:", max(scores), "| Sum:", sum(scores))

# List comprehension: build a new list in one line
doubled = [s * 2 for s in scores]
high_scores = [s for s in scores if s >= 80]
print("Doubled:", doubled)
print("Scores of 80 or more:", high_scores)
```

### Step 4: Tuples (ordered, not changeable)

```python
point = (3, 4)
x, y = point                # tuple unpacking
print("x =", x, "| y =", y, "| distance from origin =", (x ** 2 + y ** 2) ** 0.5)

try:
    point[0] = 10           # tuples cannot be changed
except TypeError as error:
    print("TypeError:", error)
```

### Step 5: Sets (no duplicates)

```python
python_class = {"Asha", "Ravi", "Meena", "Kabir"}
sql_class = {"Meena", "Kabir", "Tara"}

print("Both classes   :", sorted(python_class & sql_class))   # intersection
print("Either class   :", sorted(python_class | sql_class))   # union
print("Only Python    :", sorted(python_class - sql_class))   # difference
print("Remove duplicates from a list:", sorted(set([3, 1, 3, 2, 2, 1])))
```

### Step 6: Dictionaries (key-value pairs)

```python
student = {"name": "Asha", "age": 20, "city": "Jaipur"}
student["marks"] = 88                      # add a new key
student["age"] = 21                        # change a value
print(student)
print("City:", student["city"])
print("Phone (key missing, using get):", student.get("phone", "not available"))

for key, value in student.items():         # loop over keys and values
    print(f"  {key:<6} -> {value}")

# Nested structure: a list of dictionaries is how table rows are often stored
records = [
    {"name": "Asha", "marks": 88},
    {"name": "Ravi", "marks": 54},
    {"name": "Meena", "marks": 71},
]
print("Average marks:", sum(r["marks"] for r in records) / len(records))
```

### Step 7: Type conversion
Converting with `int()`, `float()`, `str()`, `bool()`, `list()` and `set()`. A conversion that cannot work raises an error, so we catch it with `try / except`.

```python
print(int("42") + 8, float("3.5") * 2, str(100) + " rupees")
print(int(7.9), "<- int() cuts the decimals, it does not round")
print(bool(0), bool(""), bool([]), "<- empty or zero values are False")
print(bool(5), bool("text"), bool([0]), "<- everything else is True")
print(list("data"), tuple([1, 2]), sorted(set("banana")))

for raw in ["25", "3.7", "abc"]:
    try:
        print(f"{raw!r:>6} -> {int(raw)}")
    except ValueError as error:
        print(f"{raw!r:>6} -> ValueError: {error}")
```

### Step 8: Indexing and slicing

```python
letters = ["a", "b", "c", "d", "e", "f"]
print("First:", letters[0], "| Last:", letters[-1], "| Second last:", letters[-2])
print("letters[1:4]   :", letters[1:4])      # index 1, 2, 3
print("letters[:3]    :", letters[:3])
print("letters[3:]    :", letters[3:])
print("letters[::2]   :", letters[::2])      # every second item
print("letters[::-1]  :", letters[::-1])     # reversed

word = "DataScience"
print(word[0:4], "|", word[4:], "|", word[::-1])
```

### Step 9: Practical problem 1 - count the words in a sentence
A dictionary is the natural tool for counting.

```python
sentence = "data is the new oil and data analysis turns data into insight"
word_counts = {}
for word in sentence.split():
    word_counts[word] = word_counts.get(word, 0) + 1

ranked = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
print("Most common words:", ranked[:3])
print("Unique words:", len(word_counts), "| Total words:", len(sentence.split()))
```

### Practical problem 2 - remove duplicates but keep the original order

```python
emails = ["a@x.com", "b@x.com", "a@x.com", "c@x.com", "b@x.com"]

seen = set()
unique_in_order = []
for email in emails:
    if email not in seen:
        seen.add(email)
        unique_in_order.append(email)

print("With set()           :", sorted(set(emails)), "(order is lost)")
print("Order kept (our loop):", unique_in_order)

assert unique_in_order == ["a@x.com", "b@x.com", "c@x.com"]
assert word_counts["data"] == 3 and len(word_counts) == 10
assert sum(word_counts.values()) == len(sentence.split())
print("All validation checks passed.")
```

## 11. Observations and Analysis
* A **list** is the default choice when order matters and values change; a **tuple** protects values that must not change; a **set** removes duplicates and tests membership quickly; a **dictionary** is best when values are looked up by a name.
* Slicing never raises an error for an out-of-range `stop`, but indexing a missing position (`letters[10]`) does.
* A list of dictionaries (`records`) looks just like a table. This is exactly the structure that pandas turns into a DataFrame in Assignment 5.

**Common mistakes**
| Mistake | Fix |
|---|---|
| Expecting `set()` to keep the order | Use the loop in Practical problem 2 |
| Using `int("3.7")` | Convert with `int(float("3.7"))` |
| Reading `d["key"]` for a key that may be missing | Use `d.get("key", default)` |
| Changing a list while looping over it | Loop over a copy or build a new list |
