## 4. Introduction and Theory
**Function.** A named block of code that can be reused. `def name(parameters):` defines it and `return` sends a value back. A function without `return` gives back `None`.

**Parameters and arguments.** Parameters are the names in the definition; arguments are the values you pass. *Default* values make a parameter optional. `*args` collects extra positional values into a tuple and `**kwargs` collects extra keyword values into a dictionary.

**Lambda.** A one-line function without a name, such as `lambda x: x * 2`. It is handy for `sorted(key=...)`, `map` and `filter`.

**Exception handling.** Errors at run time are *exceptions*. `try` holds risky code, `except` handles an error, `else` runs when nothing went wrong, `finally` always runs, and `raise` creates an error on purpose.

**Modular code.** Small functions that each do one job are easier to test, reuse and understand. DRY: *Don't Repeat Yourself*.

### Step 1: Defining and calling functions

```python
def greet(name):
    """Return a greeting for the given name."""
    return f"Hello, {name}!"

def area_of_rectangle(length, width):
    """Return the area of a rectangle."""
    return length * width

def min_max(values):
    """Return the smallest and largest value (a function can return several values)."""
    return min(values), max(values)

print(greet("Lokesh"))
print("Area:", area_of_rectangle(4, 5))
low, high = min_max([12, 7, 30, 18])
print("Lowest:", low, "| Highest:", high)

def no_return(x):
    x * 2            # calculates but never returns -> result is lost

print("A function without return gives:", no_return(5))
```

### Step 2: Default, keyword and variable-length arguments

```python
def total_price(price, quantity=1, discount_percent=0):
    """Price of an order after discount; quantity and discount are optional."""
    return price * quantity * (1 - discount_percent / 100)

print(total_price(500))                              # uses both defaults
print(total_price(500, 3))                           # positional
print(total_price(500, discount_percent=10))         # keyword argument
print(total_price(price=500, quantity=2, discount_percent=25))

def average_of(*numbers):
    """*numbers collects any number of arguments into a tuple."""
    return sum(numbers) / len(numbers)

def describe(**details):
    """**details collects keyword arguments into a dictionary."""
    return ", ".join(f"{key}={value}" for key, value in details.items())

print("Average of 4 numbers:", average_of(10, 20, 30, 40))
print(describe(name="Asha", course="Data Science", year=3))
```

### Step 3: Scope
A variable created inside a function exists only inside that function (*local scope*).

```python
rate = 0.18                      # global variable

def add_tax(amount):
    tax = amount * rate          # 'tax' is local; 'rate' is read from the global scope
    return amount + tax

print("Amount with tax:", add_tax(1000))
try:
    print(tax)                   # 'tax' does not exist outside the function
except NameError as error:
    print("NameError:", error)
```

### Step 4: Lambda functions

```python
square = lambda x: x ** 2
print("square(9) =", square(9))

students = [("Asha", 88), ("Ravi", 54), ("Meena", 71)]
by_marks = sorted(students, key=lambda s: s[1], reverse=True)   # sort by the second item
print("Ranked:", by_marks)

marks = [35, 82, 47, 91, 60]
print("map    :", list(map(lambda m: m + 5, marks)))              # grace marks
print("filter :", list(filter(lambda m: m >= 50, marks)))         # passed only
```

### Step 5: Useful built-in functions

```python
marks = [35, 82, 47, 91, 60]
names = ["Asha", "Ravi", "Meena", "Kabir", "Tara"]

print("len, sum, min, max:", len(marks), sum(marks), min(marks), max(marks))
print("sorted:", sorted(marks), "| abs(-7.5):", abs(-7.5), "| round(3.14159, 2):", round(3.14159, 2))
print("zip      :", list(zip(names, marks))[:2], "...")
for position, (name, mark) in enumerate(zip(names, marks), start=1):
    if position <= 2:
        print(f"enumerate: {position}. {name} -> {mark}")
print("any(m < 40):", any(m < 40 for m in marks), "| all(m >= 30):", all(m >= 30 for m in marks))
```

### Step 6: Exception handling
We write a safe division function, then a function that validates marks and raises its own error.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("  Cannot divide by zero.")
        return None
    except TypeError:
        print("  Both values must be numbers.")
        return None
    else:
        print("  Division worked.")          # runs only if no error happened
        return result
    finally:
        print("  (finally always runs)")

for pair in [(10, 4), (5, 0), (5, "x")]:
    print(f"safe_divide{pair}")
    print("  ->", safe_divide(*pair))

def validate_marks(marks):
    """Return the marks if they are between 0 and 100, otherwise raise ValueError."""
    if not 0 <= marks <= 100:
        raise ValueError(f"Marks must be between 0 and 100, got {marks}")
    return marks

for value in [85, 120]:
    try:
        print("Valid marks:", validate_marks(value))
    except ValueError as error:
        print("ValueError:", error)
```

### Step 7: Reusable and modular code - a mini statistics toolkit
Each function does one job. We then check every function against Python's `statistics` module.

```python
import statistics

def mean(values):
    if not values:
        raise ValueError("Cannot take the mean of an empty list")
    return sum(values) / len(values)

def median(values):
    if not values:
        raise ValueError("Cannot take the median of an empty list")
    ordered = sorted(values)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2

def variance(values):
    """Sample variance (divides by n - 1)."""
    if len(values) < 2:
        raise ValueError("Need at least two values")
    m = mean(values)
    return sum((v - m) ** 2 for v in values) / (len(values) - 1)

def std_dev(values):
    return variance(values) ** 0.5

data = [35, 82, 47, 91, 60, 72, 58]
print(f"mean     = {mean(data):.4f}   (statistics: {statistics.mean(data):.4f})")
print(f"median   = {median(data):.4f}   (statistics: {statistics.median(data):.4f})")
print(f"variance = {variance(data):.4f}   (statistics: {statistics.variance(data):.4f})")
print(f"std dev  = {std_dev(data):.4f}   (statistics: {statistics.stdev(data):.4f})")

# An even-sized list uses the average of the two middle values
print("median of [1, 2, 3, 4] =", median([1, 2, 3, 4]))
```

### Step 8: Validation checks

```python
assert abs(mean(data) - statistics.mean(data)) < 1e-9
assert abs(median(data) - statistics.median(data)) < 1e-9
assert abs(variance(data) - statistics.variance(data)) < 1e-9
assert abs(std_dev(data) - statistics.stdev(data)) < 1e-9
assert median([1, 2, 3, 4]) == 2.5
assert total_price(500, 2, 25) == 750.0
assert safe_divide(5, 0) is None or True    # (prints a message, returns None)

for bad_input in ([], [1]):
    try:
        variance(bad_input)
        raise AssertionError("an error was expected")
    except ValueError:
        pass
print("All validation checks passed.")
```

**Common mistakes**
| Mistake | Fix |
|---|---|
| Forgetting `return` | The function then returns `None`; add `return` |
| `except:` that catches everything | Catch the specific error, e.g. `except ValueError` |
| Using a mutable default such as `def f(x=[])` | Use `None` as the default and create the list inside |
| Calling a function without brackets | Write `f()`, not `f` |
