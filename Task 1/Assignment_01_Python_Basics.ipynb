## 4. Introduction and Theory
**Python and Data Science.** Python is a general-purpose language with simple, readable syntax. Data scientists use it because it has powerful free libraries (NumPy, pandas, Matplotlib, scikit-learn) for working with data, and because a beginner can read Python code almost like English.

**Variable.** A name that refers to a value, for example `age = 21`. Python works out the type automatically.

**Constant.** Python has no real constants. By convention a name in CAPITAL LETTERS (for example `PI`) means "do not change this value".

**Operators.**
* Arithmetic: `+ - * / // % **` (add, subtract, multiply, divide, floor-divide, remainder, power).
* Comparison: `== != < > <= >=` give `True` or `False`.
* Logical: `and`, `or`, `not` combine True/False values.

**Conditions.** `if condition:` runs a block only when the condition is True. `elif` adds more conditions and `else` handles everything left over.

**Loops.** A `for` loop repeats once for each item of a sequence. A `while` loop repeats as long as its condition stays True. `break` leaves a loop early and `continue` skips to the next round.

### Step 3: Operators

```python
a, b = 17, 5

print("Arithmetic operators")
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)      # normal division always gives a float
print("a // b =", a // b)     # floor division: whole part only
print("a % b  =", a % b)      # remainder
print("a ** b =", a ** b)     # power

print("\nComparison operators")
print("a == b:", a == b, "| a != b:", a != b, "| a > b:", a > b, "| a <= b:", a <= b)

print("\nLogical operators")
marks, attendance = 72, 80
print("Marks >= 40 and attendance >= 75:", marks >= 40 and attendance >= 75)
print("Marks >= 90 or attendance >= 75:", marks >= 90 or attendance >= 75)
print("not (marks >= 40):", not (marks >= 40))
```

### Step 4: Conditional statements
Python decides which block to run by checking the conditions from top to bottom. **Indentation (4 spaces) is part of the syntax.**

```python
def grade_for(marks):
    # Only the first condition that is True runs.
    if marks >= 75:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

for m in [92, 75, 61, 40, 39]:
    print(f"Marks {m:>3} -> Grade {grade_for(m)}")

# BMI category using the BMI value from Step 2
if bmi < 18.5:
    category = "Underweight"
elif bmi < 25:
    category = "Normal weight"
elif bmi < 30:
    category = "Overweight"
else:
    category = "Obese"
print(f"BMI {bmi:.2f} is in the category: {category}")
```

### Step 5: Loops

```python
# for loop over a list
subjects = ["Python", "Statistics", "SQL"]
for subject in subjects:
    print("Learning", subject)

# for loop with range(start, stop, step): stop is NOT included
print("Even numbers from 2 to 10:", [n for n in range(2, 11, 2)])

# while loop: how many years until 1000 rupees doubles at 8% yearly interest?
amount, years = 1000.0, 0
while amount < 2000:
    amount *= 1.08
    years += 1
print(f"It takes {years} years; the amount is then {amount:.2f}")

# break and continue
for n in range(1, 11):
    if n % 2 == 0:
        continue          # skip even numbers
    if n > 7:
        break             # stop the loop
    print("Odd number:", n)
```

### Step 6: Practical programs
**Program A: multiplication table**

```python
number = 7
for i in range(1, 11):
    print(f"{number} x {i:>2} = {number * i}")
```

**Program B: sum of digits and prime numbers**

```python
# Sum of the digits of a number
n = 9875
total, remaining = 0, n
while remaining > 0:
    total += remaining % 10     # last digit
    remaining //= 10            # remove the last digit
print(f"Sum of digits of {n} = {total}")

# Prime numbers up to 30 (a prime has no divisor between 2 and itself)
primes = []
for candidate in range(2, 31):
    is_prime = True
    for divisor in range(2, candidate):
        if candidate % divisor == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(candidate)
print("Primes up to 30:", primes)
```

**Program C: FizzBuzz (a classic interview problem)**

```python
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

**Program D: temperature conversion table (Celsius to Fahrenheit)**

```python
print("Celsius | Fahrenheit")
for celsius in range(0, 101, 25):
    fahrenheit = celsius * 9 / 5 + 32
    print(f"{celsius:>7} | {fahrenheit:>10.1f}")
```

**Program E: marks summary**

```python
marks_list = [78, 91, 55, 67, 84, 39, 72]

total = 0
highest = lowest = marks_list[0]
passed = 0
for m in marks_list:
    total += m
    if m > highest:
        highest = m
    if m < lowest:
        lowest = m
    if m >= PASS_MARK:
        passed += 1

average = total / len(marks_list)
print(f"Students: {len(marks_list)} | Average: {average:.2f}")
print(f"Highest: {highest} | Lowest: {lowest} | Passed: {passed} | Failed: {len(marks_list) - passed}")
```

## 11. Observations and Analysis
* Python needs no type declarations: the same name can hold an integer, a float or a string.
* `/` always returns a float, while `//` keeps only the whole part. Choosing the right operator avoids silent mistakes.
* A `while` loop must change something inside the loop (here `amount`), otherwise it never ends.
* Writing a small reusable `function` (`grade_for`) is cleaner than repeating `if` blocks, which prepares for Assignment 3.

**Common mistakes**
| Mistake | Fix |
|---|---|
| Using `=` where `==` is needed in a condition | `=` stores a value, `==` compares two values |
| Wrong indentation after `if` or `for` | Indent the block by 4 spaces |
| Forgetting that `range(1, 5)` stops at 4 | Use `range(1, 6)` to include 5 |
| Infinite `while` loop | Make sure the condition can become False |
